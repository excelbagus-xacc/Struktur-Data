**TUGAS 1**

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

# Root = jumlah huruf nama lengkap
t_root = Node(20)
# Anak kiri root = jumlah huruf vokal
t_root.left = Node(7)
# Anak kanan root = dua digit terakhir NPM
t_root.right = Node(23)
# Anak kiri dari node kiri = tanggal lahir (DD)
t_root.left.left = Node(15)
# Anak kanan dari node kanan = bulan lahir (MM)
t_root.right.right = Node(8)
# Fungsi inorder traversal
def inorder(node):
    if node:
        inorder(node.left)
        print(node.data, end=' ')
        inorder(node.right)
print("Inorder traversal Tugas 1:")
inorder(t_root)
```

**TUGAS 2**

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

class BinarySearchTree:
    def __init__(self):
        self.root = None
    # Menyisipkan data sesuai aturan BST (kecil ke kiri, besar ke kanan)
    def insert(self, root, data):
        if root is None:
            return Node(data)
        if data < root.data:
            root.left = self.insert(root.left, data)
        else:
            root.right = self.insert(root.right, data)
        return root

# 7 angka unik berdasarkan identitas diri
data_list = [156, 30, 5, 69, 120, 7, 97]
bst = BinarySearchTree()
root = None
for value in data_list:
    root = bst.insert(root, value)

# Inorder: hasilkan data terurut
def inorder(node):
    if node:
        inorder(node.left)
        print(node.data, end=' ')
        inorder(node.right)

# Preorder: cetak struktur dari root ke bawah
def preorder(node):
    if node:
        print(node.data, end=' ')
        preorder(node.left)
        preorder(node.right)

# Postorder: proses subtree sebelum root
def postorder(node):
    if node:
        postorder(node.left)
        postorder(node.right)
        print(node.data, end=' ')

print("Inorder Traversal:")
inorder(root)
print("\nPreorder Traversal:")
preorder(root)
print("\nPostorder Traversal:")
postorder(root)

# Fitur pencarian
def search(node, key):
    if node is None or node.data == key:
        return node
    if key < node.data:
        return search(node.left, key)
    return search(node.right, key)

# Cari dua digit terakhir NPM (56) - tidak ada dalam list
key = 56
result = search(root, key)
print(f"\n\nCari {key}: {'ditemukan dalam tree.' if result else 'tidak ditemukan.'}")

# Cari angka yang tidak ada dalam list
key = 77
result = search(root, key)
print(f"Cari {key}: {'ditemukan dalam tree.' if result else 'tidak ditemukan.'}")
```
