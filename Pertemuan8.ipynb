Representasi Graf dengan Adjacency List

```python
# Membuat graf tak berarah menggunakan adjacency list
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A','F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

# Menampilkan graf
def print_graph (graf) :
  for node in graf:
    print (f" {node} -> {graf[node]} ")

print_graph(graph)
```

Traversal Graf menggunakan BFS

```python
from collections import deque

def bfs(graph, start):
    visited = set()
    queue = deque([start])

    while queue:
        vertex = queue.popleft()
        if vertex not in visited:
            print(vertex, end=" ")
            visited.add(vertex)
            queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])

# Contoh pemanggilan
bfs(graph, 'A')
```

Traversal Graf menggunakan DFS

```python
def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()

    if start not in visited:
        print(start, end=" ")
        visited.add(start)
        for neighbor in graph[start]:
            dfs(graph, neighbor, visited)

# Contoh pemanggilan
dfs(graph, 'A')
```

Latihan Personal

```python
def count_neighbors(graph):
    result = {}
    for node in graph:
        result[node] = len(graph[node])
    return result

graph_modul = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A','F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

neighbors_count = count_neighbors(graph_modul)
print('Jumlah tetangga untuk setiap simpul:')
for node, count in neighbors_count.items():
  print(f'Simpul {node} memiliki {count} tetangga')
```

2. BFS yang Mengembalikan List

```python
from collections import deque

# Membuat graf tak berarah menggunakan adjacency list
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A','F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

def bfs_return(graph, start):
    visited = set()
    queue = deque([start])
    order = []

    while queue:
        vertex = queue.popleft()
        if vertex not in visited:
            order.append(vertex)
            visited.add(vertex)
            queue.extend([n for n in graph[vertex] if n not in visited])

    return order

result = bfs_return(graph, 'A')
print("Urutan BFS:", result)
```

3. Fungsi find_path() menggunakan DFS

```python
def find_path(graph, start, end, visited=None, path=None):
    if visited is None:
        visited = set()
    if path is None:
        path = []

    visited.add(start)
    path = path + [start]

    if start == end:
        return path

    for neighbor in graph[start]:
        if neighbor not in visited:
            result = find_path(graph, neighbor, end, visited, path)
            if result:
                return result

    return None

jalur = find_path(graph, 'A', 'F')
print("Jalur dari A ke F:", jalur)
# Output: ['A', 'C', 'F']
```

4. Fungsi is_connected()

```python
def is_connected(graph):
    if not graph:
        return True

    # Ambil satu simpul sembarang sebagai titik awal
    start = next(iter(graph))

    # Lakukan DFS dari simpul tersebut
    visited = set()
    stack = [start]

    while stack:
        node = stack.pop()
        if node not in visited:
            visited.add(node)
            stack.extend([n for n in graph[node] if n not in visited])

    # Jika semua simpul berhasil dikunjungi, graf terhubung
    return len(visited) == len(graph)

print("Graf terhubung?", is_connected(graph))         # True
print("Graf kota terhubung?", is_connected(graph_kota))  # True

# Contoh graf tidak terhubung
graph_putus = {
    'A': ['B'],
    'B': ['A'],
    'C': ['D'],
    'D': ['C']
}
print("Graf putus terhubung?", is_connected(graph_putus))  # False
```
